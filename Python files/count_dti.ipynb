```python
import os

def count_dwi_folders(root_dir):
    count = 0
    for dirpath, dirnames, _ in os.walk(root_dir):

        if 'dwi' in dirnames:
            count += 1
    return count

def count_func_folders(root_dir):
    count = 0
    for dirpath, dirnames, _ in os.walk(root_dir):

        if 'func' in dirnames:
            count += 1
    return count
    

count_dwi_folders("../../Storage Repository/")
```

```python
def count_subjects_with_dwi(root_dir):
    subjects_with_dwi = set()

    for dirpath, dirnames, _ in os.walk(root_dir):
        # Check if current folder contains 'dwi' subfolder
        if 'dwi' in dirnames:
            # Extract subject folder from the path
            # Assumption: subject folder name contains 'sub-' and is in the path
            parts = dirpath.split(os.sep)
            subject_folders = [p for p in parts if p.startswith('sub-')]
            if subject_folders:
                subjects_with_dwi.add(subject_folders[0])

    return len(subjects_with_dwi)
    
count_subjects_with_dwi("../../Storage Repository/")
```
