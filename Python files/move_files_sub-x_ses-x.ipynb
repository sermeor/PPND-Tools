```python
import os

def get_files(directory):
    """
    Iterates through files in a directory, checks for 'TRFEP', and prompts for confirmation before moving.

    Args:
        directory (str): The directory to search for files.

    Returns:
        None
    """

    for root, dirs, files in os.walk(directory):
        for file in files:
            if 'TRFEP' in file:
                full_path = os.path.join(root, file)
                splits = file.split('_')
                sub = splits[2]
                ses = splits[3]
                move_dir = os.path.join('TRFEP_BIDS', 'sub-'+sub, 'ses-'+ses[0], 'func')

                print(f"Found file: {full_path}")
                print(f"Target directory: {move_dir}")

                confirmation = input("Move file? (1/0): ")
                if confirmation.lower() == '1':
                    try:
                        os.rename(full_path, os.path.join(move_dir, file))
                        print(f"Successfully moved {full_path} to {move_dir}")
                    except OSError as e:
                        print(f"Error moving file: {e}")
                else:
                    print("Skipping file.")

if __name__ == "__main__":
    directory = input("Enter directory to search: ")
    get_files(directory)
```
