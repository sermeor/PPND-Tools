```python
import os

def rename_files_with_ids(directory):
    for root, dirs, files in os.walk(directory):
        for filename in files:
            if filename.endswith('.json') or filename.endswith('.nii.gz') or filename.endswith('.nii') or filename.endswith('.bval') or filename.endswith('.bvec'):

                subject_id = root.split(os.path.sep)[1]
                session_id = root.split(os.path.sep)[2]
                new_filename = f"{subject_id}_{session_id}_{filename}"
                os.rename(os.path.join(root, filename), os.path.join(root, new_filename))
                print(subject_id)
                print(session_id)
                print(filename)
                print(new_filename)
                
# Example usage:
directory = 'temp_CARDS/data'
rename_files_with_ids(directory)
```
