```python
import os
import shutil
import glob

def organize_all_subjects(root_folder, session_name="ses-A"):
    """
    Organize all sub-* folders in the root folder by moving their contents into a session folder.

    Parameters:
    - root_folder (str): Path to the main dataset folder containing sub-* folders.
    - session_name (str): Name of the session folder to create inside each subject (default: "ses-A").
    """
    # Find all sub-* folders
    sub_folders = glob.glob(os.path.join(root_folder, "sub-*"))
    
    if not sub_folders:
        print("No sub-* folders found.")
        return
    
    for sub_folder in sub_folders:
        session_folder = os.path.join(sub_folder, session_name)
        os.makedirs(session_folder, exist_ok=True)
        
        for item in os.listdir(sub_folder):
            item_path = os.path.join(sub_folder, item)
            # Skip the session folder itself
            if item_path == session_folder:
                continue
            shutil.move(item_path, session_folder)
        
        print(f"Organized {sub_folder} -> {session_folder}")

# Example usage:
organize_all_subjects(r"Storage Repository/SRPBS/data")
```
