```python
import os

def get_all_formats(directory):
    formats = set()
    for root, dirs, files in os.walk(directory):
        for file in files:
            _, extension = os.path.splitext(file)
            if extension:
                formats.add(extension)
    return formats

dir = "temp_CARDS/data"
all_formats = get_all_formats(dir)
print("All possible formats in the directory:")
print(all_formats)
```
