```python
import os
import filecmp

def compare_directories(dir1, dir2):
    """
    Compare two directories recursively and return a dict of differences.

    Returns a dict with keys:
    - 'only_in_dir1': list of relative file paths only in dir1
    - 'only_in_dir2': list of relative file paths only in dir2
    - 'differ': list of relative file paths that exist in both but differ
    """
    differences = {
        'only_in_dir1': [],
        'only_in_dir2': [],
        'differ': []
    }

    # Helper to get relative paths of all files in a directory recursively
    def get_all_files(base_dir):
        all_files = []
        for root, _, files in os.walk(base_dir):
            for f in files:
                full_path = os.path.join(root, f)
                rel_path = os.path.relpath(full_path, base_dir)
                all_files.append(rel_path)
        return set(all_files)

    files_dir1 = get_all_files(dir1)
    files_dir2 = get_all_files(dir2)

    # Files only in one directory
    differences['only_in_dir1'] = sorted(list(files_dir1 - files_dir2))
    differences['only_in_dir2'] = sorted(list(files_dir2 - files_dir1))



    return differences


# Example usage:
if __name__ == "__main__":
    dir1 = r"C:\Users\sergi\King's College London\Department of Psychosis Shared Database Initiative - Documents\Storage Repository\BGSCHZ"
    dir2 = r"Z:\BGSCHZ"

    diffs = compare_directories(dir1, dir2)
    print("Only in dir1:", diffs['only_in_dir1'])
    print("Only in dir2:", diffs['only_in_dir2'])
```
