```python
import os
import shutil

def remove_files_and_folders_matching_pattern(root_folder, pattern):
    for foldername, subfolders, filenames in os.walk(root_folder, topdown=False):
        for filename in filenames:
            if pattern in filename:
                file_path = os.path.join(foldername, filename)
                #os.remove(file_path)
                print(f"Removed file: {file_path}")
        for subfolder in subfolders:
            if pattern in subfolder:
                folder_path = os.path.join(foldername, subfolder)
                shutil.rmtree(folder_path)
                print(f"Removed folder: {folder_path}")

# Example usage:
root_folder = 'temp_CARDS/data'
pattern = 'func'
remove_files_and_folders_matching_pattern(root_folder, pattern)
```
