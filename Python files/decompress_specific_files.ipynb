```python
import os
import gzip
import shutil

def decompress_gzip(directory):
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith(".gz") and ('ADNI' in file or 'T2' in file):
                file_path = os.path.join(root, file)
                print("Decompressing:", file_path)
                # Output file name will be the same as input file but without '.gz' extension
                output_file_path = os.path.splitext(file_path)[0]
                try:
                    with gzip.open(file_path, 'rb') as f_in, open(output_file_path, 'wb') as f_out:
                        shutil.copyfileobj(f_in, f_out)
                    print("Extraction complete.")
                except Exception as e:
                    print('---------------')
                    print(f"Error decompressing {file_path}: {e}")
                    print('---------------')

# Example usage:
directory_to_search = "TRFEP"
decompress_gzip(directory_to_search)
```
