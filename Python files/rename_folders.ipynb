```python
import os

def rename_session_folders(root_dir):
    for subdir, dirs, files in os.walk(root_dir):
        for dir_name in dirs:
            if dir_name.startswith("session-"):
                old_path = os.path.join(subdir, dir_name)
                new_name = dir_name.replace("session-", "ses-")
                new_path = os.path.join(subdir, new_name)
                os.rename(old_path, new_path)

# Example usage:
root_directory = "TRFEP_BIDS"
rename_session_folders(root_directory)
```
