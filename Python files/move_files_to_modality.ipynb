```python
import os
import shutil

def get_files(directory):
    for root, dirs, files in os.walk(directory):
        for file in files:
            if "t1" in file.lower():
                file_path = os.path.join(root, file)
                target_folder = os.path.join(root, "anat")
                try:
                    shutil.move(file_path, target_folder)
                except:
                    print('Error on' + file_path)
# Example usage:
directory = 'temp_CARDS/data'
get_files(directory)
```
