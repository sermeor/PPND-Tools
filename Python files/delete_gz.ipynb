```python
import os
import gzip
import shutil

def decompress_gzip(directory):
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith(".gz") and ('sub-' in file ):
                file_path = os.path.join(root, file)
                try:
                    os.remove(file_path)
                    print("Removing:", file_path)
                except:
                    print('---------------')
                    print(f"Error decompressing:", file_path)
                    print('---------------')


# Example usage:
directory_to_search = "../BCSPS"
decompress_gzip(directory_to_search)
```
