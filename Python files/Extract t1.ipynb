```python
import os
import shutil

def search_and_copy_files(search_path, search_string, destination_folder):
    """
    Search for files containing a specific string within directories and subdirectories,
    then copy those files to a destination folder.

    Parameters:
        search_path (str): The root directory to start the search.
        search_string (str): The string to search for within files.
        destination_folder (str): The folder where matching files will be copied.
    """
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)

    for root, _, files in os.walk(search_path):
        for file in files:
            file_path = os.path.join(root, file)
            try:
                with open(file_path, 'r', errors='ignore') as f:
                    if search_string in f.read():
                        dest_path = os.path.join(destination_folder, file)

                        # Handle duplicate file names
                        if os.path.exists(dest_path):
                            base, ext = os.path.splitext(file)
                            counter = 1
                            while os.path.exists(dest_path):
                                dest_path = os.path.join(destination_folder, f"{base}_{counter}{ext}")
                                counter += 1

                        shutil.copy2(file_path, dest_path)
                        print(f"Copied: {file_path} -> {dest_path}")
            except Exception as e:
                print(f"Error reading {file_path}: {e}")

# Example usage
search_and_copy_files("MOA/", "T1w.nii", "t1files/")
```
