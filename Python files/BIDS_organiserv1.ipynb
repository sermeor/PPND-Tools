```python
import os
import re
import shutil

def get_files(directory):
    file_list = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            file_list.append(os.path.join(root, file))
    return file_list

def extract_patient_session(file_path):
    # Define the regex pattern to match patient ID and session
    pattern = r'TRFEP\\(TRFEP\d+)([A-Z])\\'
    match = re.search(pattern, file_path)
    if match:
        patient_id = match.group(1)
        session = match.group(2)
        return patient_id, session
    else:
        return None, None


data_dir = "TRFEP/Images/TRFEP"
destination_dir = 'TRFEP_BIDS'

list_of_files = get_files(data_dir)

for file in list_of_files:
    subject_id, session = extract_patient_session(file)
    if (subject_id == None or session == None):
        print('Error on: ' + file)
    else:
        destination_subdir = os.path.join(destination_dir, 'sub-' + subject_id, 'ses-' + session)
        os.makedirs(destination_subdir, exist_ok=True)
        shutil.copy(file, destination_subdir)
    
    

```
