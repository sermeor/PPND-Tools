```python
import os
import subprocess
import shutil



def convert_dicom_to_nifti(input_dir, output_dir):
    # Command to run dcm2niix
    command = ["dcm2niix", "-o", output_dir, input_dir]

    try:
        # Execute the command
        subprocess.run(command, check=True)
    except subprocess.CalledProcessError as e:
        # An error occurred while executing the command
        print('==============================================')
        print("Conversion failed:", e)
        print("In file: " + input_dir)
        print('==============================================')


def convert_to_nifti_directory(directory):
    for root, dirs, files in os.walk(directory):
        for dir_name in dirs:
            dir_path = os.path.join(root, dir_name)
            # Get the relative path of the directory
            rel_path = os.path.relpath(dir_path, directory)
            # Count the number of directories in the relative path to determine the level
            level = rel_path.count(os.sep)
            if level == 1:
                convert_dicom_to_nifti(dir_path, os.path.dirname(dir_path))


def delete_level_1_folders(directory):
    for root, dirs, files in os.walk(directory, topdown=False):
        for dir_name in dirs:
            dir_path = os.path.join(root, dir_name)
            # Get the relative path of the directory
            rel_path = os.path.relpath(dir_path, directory)
            # Count the number of directories in the relative path to determine the level
            level = rel_path.count(os.sep)
            if level == 1:
                print("Deleting:", dir_path)
                # Delete the directory and its contents
                shutil.rmtree(dir_path)
                

def rename_folders(directory):
    for root, dirs, files in os.walk(directory):
        for dir_name in dirs:
            dir_path = os.path.join(root, dir_name)
            # Get the relative path of the directory
            rel_path = os.path.relpath(dir_path, directory)
            # Count the number of directories in the relative path to determine the level
            level = rel_path.count(os.sep)
            # Rename the directory
            new_name = os.path.join(root, f"sub-{dir_name}")
            os.rename(dir_path, new_name)
            print(f"Renamed {dir_path} to {new_name}")


def create_folders(directory):
    # List of folders to create
    folders_to_create = ["mrs", "anat", "func"]
    
    # Walk through the directory tree
    for root, dirs, files in os.walk(directory):
        # Iterate over each directory
        for dir_name in dirs:
            dir_path = os.path.join(root, dir_name)
            # Create the subfolders within the directory
            for folder_name in folders_to_create:
                folder_path = os.path.join(dir_path, folder_name)
                os.makedirs(folder_path, exist_ok=True)
                print(f"Created folder: {folder_path}")




def remove_folders(directory):
    # List of folders to remove
    folders_to_remove = ["mrs", "anat", "func"]
    
    # Walk through the directory tree
    for root, dirs, files in os.walk(directory, topdown=False):  # Use topdown=False to ensure we remove subdirectories first
        # Iterate over each directory
        for dir_name in dirs:
            dir_path = os.path.join(root, dir_name)
            # Check if the subdirectory is one of the folders to remove
            if dir_name in folders_to_remove:
                shutil.rmtree(dir_path)
                print(f"Removed folder: {dir_path}")

# Example usage
# remove_folders("/path/to/your/directory")

dir = "temp_CARDS"
convert_to_nifti_directory(dir)


```
