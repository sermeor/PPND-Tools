```python
import os

def get_files(directory):
    file_list = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if 'TreatFep' in file:
                file_name, format_name = file.split('.')
                splits = file_name.split('_')
                new_file = 'sub-'+ splits[2] + '_ses-' + splits[3]+ '_'+splits[0]+'_'+splits[1]
                if len(splits)>4:
                    new_file = new_file +'_'+splits[4]
                if len(splits)>5:
                    new_file = new_file +'_'+splits[5]
                if len(splits)>6:
                    new_file = new_file +'_'+splits[6]
                
                try:
                    new_file_path = os.path.join(root, new_file + '.' +format_name)
                    os.rename(os.path.join(root, file), new_file_path)
                    file_list.append(new_file_path)
                except Exception as e:
                    print(f"Error renaming file {file}: {e}")
    
    return file_list

dir = "TRFEP"
list_of_files = get_files(dir)
```
