```python
import os 

def rename_items(root_directory):
    for root, dirs, files in os.walk(root_directory, topdown=False):
        for name in dirs + files:
            item_path = os.path.join(root, name)
            
            if name.startswith("sub-") and not name.startswith("sub-MUNI"):
                new_item_name = name.replace("sub-", "sub-MUNI", 1)
                new_item_path = os.path.join(root, new_item_name)
                             
                try:
                    os.rename(item_path, new_item_path)
                    print(f"Renamed {item_path} to {new_item_path}")
                except Exception as e:
                    print(f"Error renaming {item_path}: {e}")


def rename_items_inside(root_directory):
    """
    Renames files and directories within root_directory that contain 'sub-' 
    (but not 'sub-MUNI') anywhere in their name.
    Replaces the first occurrence of 'sub-' with 'sub-MUNI'.
    """
    for root, dirs, files in os.walk(root_directory, topdown=False):
        for name in dirs + files:
            # Skip names that already have sub-MUNI
            if "sub-" in name and "sub-MUNI" not in name:
                old_path = os.path.join(root, name)
                # Replace only the first occurrence
                new_name = name.replace("sub-", "sub-MUNI", 1)
                new_path = os.path.join(root, new_name)
                
                try:
                    os.rename(old_path, new_path)
                    print(f"Renamed {old_path} → {new_path}")
                except Exception as e:
                    print(f"Error renaming {old_path}: {e}")

# Example usage
root_directory = '../../Storage Repository/MUNI'
rename_items_inside(root_directory)                    
            

```
