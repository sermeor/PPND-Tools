```python
import os

def get_files(directory):
    file_list = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if ('T2' in file or 'ADNI' in file) and (file.endswith('.nii') and file.startswith('sub-')):
                print(file)
                os.remove(os.path.join(root, file))
                
    for root, dirs, files in os.walk(directory):
        for file in files:
            if ('T2' in file or 'ADNI' in file) and (file.endswith('.nii') and file.startswith('anon_')):
                print(file)
                old_name = os.path.join(root, file)
                new_name = os.path.join(root, file.replace('anon_', ''))
                os.rename(old_name, new_name)



dir = "TRFEP"

get_files(dir)
```
