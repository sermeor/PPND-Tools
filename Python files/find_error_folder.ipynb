```python
import os
import shutil



def find_err_dirs(root_path):
    """Find and print all 'err' directories."""
    for dirpath, dirnames, _ in os.walk(root_path):
        if 'err' in dirnames:
            print(os.path.join(dirpath, 'err'))

def delete_err_dirs(root_path):
    """Delete all 'err' directories under the root path safely. Skips on error."""
    for dirpath, dirnames, _ in os.walk(root_path):
        if 'err' in dirnames:
            err_path = os.path.join(dirpath, 'err')
            print(f"Deleting: {err_path}")
            try:
                shutil.rmtree(err_path)
            except Exception as e:
                print(f"Failed to delete {err_path}: {e}")

# Example usage:
#find_err_dirs('../Storage Repository/MUNI/')
delete_err_dirs('../Storage Repository/MUNI/')
```
