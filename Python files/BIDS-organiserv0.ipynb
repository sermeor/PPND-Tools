```python
import os
import shutil

# Define the source directories
reports_dir = 'Jana/NKI-RS/report'
labels_dir = 'Jana/NKI-RS/label'
mri_dir = 'Jana/NKI-RS/mri'

# Define the destination directory
destination_dir = 'BIDS'

# Function to extract subject ID and session from file path
def extract_info(file_path):
    filename = os.path.splitext(os.path.basename(file_path))[0]
    parts = filename.split('_')
    subject_id = None
    session = None
    for part in parts:
        if 'sub-' in part:
            subject_id = part.split('-')[1]
        elif 'ses-' in part:
            session = part.split('-')[1]
    return subject_id, session

# Create the destination directory if it doesn't exist
os.makedirs(destination_dir, exist_ok=True)

# Iterate over each file in MRI directory
for mri_file in os.listdir(mri_dir):
    subject_id, session = extract_info(mri_file)
    if (subject_id == None or session == None):
        print('Error on: ' + mri_file)
    else:
        destination_subdir = os.path.join(destination_dir, 'sub-' + subject_id, 'session-' + session, 'anat')
        os.makedirs(destination_subdir, exist_ok=True)
        shutil.copy(os.path.join(mri_dir, mri_file), destination_subdir)

# Iterate over each file in reports directory
for report_file in os.listdir(reports_dir):
    subject_id, session = extract_info(report_file)
    if (subject_id == None or session == None):
        print('Error on: ' + report_file)
    else:
        destination_subdir = os.path.join(destination_dir, 'sub-' + subject_id, 'session-' + session, 'anat')
        os.makedirs(destination_subdir, exist_ok=True)
        shutil.copy(os.path.join(reports_dir, report_file), destination_subdir)

# Iterate over each file in labels directory
for label_file in os.listdir(labels_dir):
    subject_id, session = extract_info(label_file)
    if (subject_id == None or session == None):
        print('Error on: ' + label_file)
    else:
        destination_subdir = os.path.join(destination_dir, 'sub-' + subject_id, 'session-' + session, 'anat')
        os.makedirs(destination_subdir, exist_ok=True)
        shutil.copy(os.path.join(labels_dir, label_file), destination_subdir)
```
