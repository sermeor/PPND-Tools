```python
import os
import re

def standardize_filenames(directory):
    # Iterate through files in the directory
    for filename in os.listdir(directory):
        # Check if the file is a regular file
        if os.path.isfile(os.path.join(directory, filename)):
            # Define the regex pattern to match the desired format
            pattern = r'sub-\w+_ses-\w+_(.*)'
            # Check if the filename matches the desired format
            match = re.match(pattern, filename)
            if not match:
                print(filename)
                # Extract the filename without extension
                file_name_no_ext = os.path.splitext(filename)[0]
                # Extract the prefix (e.g., 'TreatFep_Resting_TRFEP01_A')
                prefix_match = re.match(r'(.*_TRFEP\d+_A)_?(.*)', file_name_no_ext)
                if prefix_match:
                    prefix = prefix_match.group(1)
                    suffix = prefix_match.group(2)
                    # Construct the new filename
                    new_filename = f'sub-{prefix}_ses-A_{suffix}.txt'
                    # Ask for user confirmation
                    print(f"Change {filename} to {new_filename}? (y/n)")
                    choice = input().lower()
                    if choice == 'y':
                        # Perform the renaming
                        os.rename(os.path.join(directory, filename), os.path.join(directory, new_filename))
                        print(f'Renamed {filename} to {new_filename}')
                    else:
                        print(f'Skipped renaming {filename}')

# Example usage:
directory_path = 'TRFEP'
standardize_filenames(directory_path)
```
