```python
import os

def get_files(directory):
    file_list = []
    for root, dirs, files in os.walk(directory):
        # Check if the current directory is a session folder (e.g., "ses-X")
        if os.path.basename(root).startswith("ses-"):
            for file in files:
                file_list.append(os.path.join(root, file))
                print(os.path.join(root, file))
    return file_list

dir = "TRFEP"
list_of_files = get_files(dir)
```
