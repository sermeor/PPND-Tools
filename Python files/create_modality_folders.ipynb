```python
import os

def create_session_folders(root_folder):
    for root, dirs, files in os.walk(root_folder):
        for dir_name in dirs:
            if dir_name.startswith("ses-"):
                session_path = os.path.join(root, dir_name)
                anat_path = os.path.join(session_path, "anat")
                try:
                    os.makedirs(anat_path, exist_ok=True)
                    print(f"Created 'anat' folder in {session_path}")
                except Exception as e:
                    print(f"Error creating 'anat' folder in {session_path}: {e}")
                    


# Example usage:
root_folder = 'temp_CARDS/data'
create_session_folders(root_folder)
```
