```python
import os
import shutil

# Define directories
base_dir = "../Storage Repository/SRPBS/"
label_dir = os.path.join(base_dir, "label")
remaining_dir = os.path.join(base_dir, "T1w")

# Ensure the remaining directory exists
os.makedirs(remaining_dir, exist_ok=True)

# Get all .nii files in the base directory (excluding "label/")
nii_files = [f for f in os.listdir(base_dir) if (f.endswith(".nii") and f.startswith("sub-"))]

# Get all corresponding "catROI_sub-" files in the label directory
catroi_files = {f.replace("catROI_", "").replace(".mat", "") for f in os.listdir(label_dir) if f.startswith("catROI_")}

# Find .nii files without a corresponding "catROI_" file
unmatched_files = [f for f in nii_files if f.replace(".nii", "") not in catroi_files]

# Copy unmatched files to "remaining/" folder
for file in unmatched_files:
    src = os.path.join(base_dir, file)
    dst = os.path.join(remaining_dir, file)
    shutil.copy(src, dst)
    print(f"Copied: {file} → {remaining_dir}")
```
