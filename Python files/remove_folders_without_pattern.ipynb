```python
import os

def remove_unwanted_folders(root_dir):
    for root, dirs, files in os.walk(root_dir, topdown=False):
        for name in dirs:
            if not (name.startswith('sub-') or name.startswith('session-')):
                folder_path = os.path.join(root, name)
                for f in os.listdir(folder_path):
                    file_path = os.path.join(folder_path, f)
                    os.remove(file_path)
                os.rmdir(folder_path)

# Replace 'temp_TRFEP/TRFEP_BIDS' with your actual root directory
root_directory = 'temp_TRFEP/TRFEP_BIDS'
remove_unwanted_folders(root_directory)
```
