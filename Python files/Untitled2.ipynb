```python
import os
import shutil

def move_files_to_ses_x(directory):
    for root, dirs, files in os.walk(directory):
        # Check if we are in the nested anat/anat/anat folder
        if 'anat' in root.split(os.sep)[-3:]:
            for file in files:
                # Construct the full file path
                file_path = os.path.join(root, file)
                # Determine the ses-A folder
                ses_a_folder = os.path.abspath(os.path.join(root, '../../..'))
                # Move the file to the ses-A folder
                ses_a_folder = os.path.abspath(os.path.join(root.split('ses-A')[0], 'ses-A'))
                shutil.move(file_path, ses_a_folder)
                print(f"Moved {file_path} to {ses_a_folder}")

directory = 'temp_CARDS/data'
move_files_to_ses_x(directory)
```
