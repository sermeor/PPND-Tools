```python
import os
import shutil

def get_files(directory, match, modality):
    for root, dirs, files in os.walk(directory):
        for file in files:
            if match in file:
                prev_folder = root.split(os.path.sep)[-1]
                if 'ses' in prev_folder:
                    # Get the modality folder path
                    modality_folder = os.path.join(root, modality)
                    # Move the file to the modality folder
                    file_path = os.path.join(root, file)
                    shutil.move(file_path, modality_folder)
                    #print(f"Moved {file} to {modality_folder}")
                    print(file_path)

dir = "TRFEP"
match = 'SAT'
modality = 'extra_data'
get_files(dir, match, modality)
```
